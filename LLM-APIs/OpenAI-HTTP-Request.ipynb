Coding HTTP Request along with Payload without using OpenAI Client Library Endpoint

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv(override=True)

#OpenAI Model
openai_api_key = os.getenv("OPENAI_API_KEY")
if not openai_api_key:
    print("No API key found")
elif not openai_api_key.startswith("sk-proj"):
    print("Invalid API Key")
else:
    print("Valid API Key")

#Gemini Model
gemini_api_key = os.getenv("GEMINI_API_KEY")
if not gemini_api_key:
    print("No API key found")
elif not gemini_api_key.startswith("AI"):
    print("Invalid API Key")
else:
    print("Valid API Key")
```

```python
system_prompt = """ You are a helpful assistant that can answer questions and help with tasks. """
user_prompt = "Capital city of maharashtra?"
messages = [
    {"role":"system", "content":system_prompt},
    {"role":"user", "content":user_prompt}
]
```

```python
openai_url="https://api.openai.com/v1/chat/completions"
gemini_url="https://generativelanguage.googleapis.com/v1beta/openai/chat/completions"

openai_model="gpt-5-nano"
gemini_model="gemini-3-flash-preview"
```

```python
header = {
    "Authorization": f"Bearer {gemini_api_key}",
    "Content-Type": "application/json"
}

payload = {
    "model":gemini_model,
    "messages":messages
}

response = requests.post(
    url=gemini_url,
    headers=header,
    json=payload
)

response.json()
```
